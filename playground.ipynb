```python
import json
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer("/home/xingkun/ant/outputs/checkpoints/constraint-encoder-v1")

with open("/home/xingkun/ant/data/processed/val_triplets.jsonl", "r", encoding="utf-8") as f:
    row = json.loads(next(f))
```

```python
s1 = "Two woman are holding packages."
s2 = "The men are fighting outside a deli."

emb = model.encode([s1, s2], normalize_embeddings=True)
sim = float(np.dot(emb[0], emb[1]))  # cosine, because normalized

print("sim =", sim)
```

```python
q = "Find hotels that are not dirty"
p = "The hotel is clean and hygienic"
n = "The hotel is dirty and unhygienic"
emb = model.encode([q, p, n], normalize_embeddings=True)
sim_qp = float(np.dot(emb[0], emb[1]))
sim_qn = float(np.dot(emb[0], emb[2]))

print("query:", q)
print("sim(q, pos) =", sim_qp)
print("sim(q, neg) =", sim_qn)
print("pass:", sim_qp > sim_qn)
```

```python
q = "Find hotels that are not dirty"
pos = "The hotel is clean and hygienic with spotless rooms."
neg = "Guests report dirty bathrooms and poor hygiene."

emb_q_pos = model.encode([q, pos], normalize_embeddings=True)
emb_q_neg = model.encode([q, neg], normalize_embeddings=True)

sim_qp = float(np.dot(emb_q_pos[0], emb_q_pos[1]))
sim_qn = float(np.dot(emb_q_neg[0], emb_q_neg[1]))

print(sim_qp, sim_qn, sim_qp > sim_qn)
```
